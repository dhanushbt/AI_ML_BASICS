#Implementaion of Vector using numpy array


```python
import numpy as np
```

```python
array_V=np.array([1,0,-1])
array_W=np.array([0,-1,1])
```

Vector addtion and deletion

```python
vec_add=array_V+array_W
print("Vector addtion",vec_add)
vec_del=array_V-array_W
print("Vector deletion",vec_del)
```

Scaler multiplication

```python
Scaler_val=3
print("Scaler multiplication",Scaler_val*array_V)
```

Dot Product

```python
dot_product=np.dot(array_V,array_W)
print("Dot product",dot_product)
```

Length of Vector

```python
len_v=np.linalg.norm(array_V)
print("Length of Vector",len_v)
```
