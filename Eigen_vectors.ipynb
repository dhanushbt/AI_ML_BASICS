```python
import numpy as np
```

```python
a=np.array([[2,-2,3],[1,1,1],[1,3,-1]])
a
```

Eigen value and vectors

> Add blockquote



```python
values,vector=np.linalg.eig(a)
print("value:",values)
print("vector:",vector)
```

```python
b=np.array([[2,0,0],[0,1,0],[0,0,3]])
```

```python
values1,vector1=np.linalg.eig(b)
print("value:",values1)
print("vector:",vector1)
```
