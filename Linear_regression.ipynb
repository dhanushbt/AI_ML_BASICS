Linear Regression model

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import fetch_california_housing
dataset= pd.DataFrame(fetch_california_housing().data, columns=fetch_california_housing().feature_names)
```

```python
dataset.describe()
```

```python
targets= fetch_california_housing().target
```

```python
fetch_california_housing().feature_names
```

```python
dataset.head()
```

```python
dataset.plot(x='MedInc', y='target', style='o')
plt.show()
```

```python
x=dataset.iloc[:,:-1].values
y=dataset.iloc[:,-1].values
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test= train_test_split(x,y, test_size=0.2, random_state=0)
```

```python
from sklearn.linear_model import LinearRegression
regressor= LinearRegression()
regressor.fit(x_train,y_train)
```

```python
print(regressor.intercept_)
```

```python
print(regressor.coef_)
```

```python
y_prd= regressor.predict(x_test)
```

```python
final_vals= pd.DataFrame({'Actual': y_test, 'Predicted': y_prd})
```

```python
print(final_vals)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
print(r2_score(y_test,y_prd))
print(mean_absolute_error(y_test,y_prd))
print(mean_squared_error(y_test,y_prd))
```
