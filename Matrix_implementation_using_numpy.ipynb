#Matrix implementation

```python
import numpy as np
```

```python
p=np.array([[1,2,3],[4,5,6]])
q=np.array([[1,2],[3,4],[5,6]])
r=np.array([[1,0,1],[2,0,1]])
a=np.array([[1,9,3],[4,5,6],[7,8,9]])
```

Matrix addition

```python
s=np.add(p,r)
print(s)
```

Matrix substraction

```python
t=np.subtract(p,r)
print(t)
```

matrix multiplication

```python
d=np.linalg.multi_dot([p,q])
print(d)
```

Note for 1d array it does dot product of vector
But for 2d array it does matrix multiplication




```python
d=p @ q
print(d)
```

Determinant of Matrix

```python
det=np.linalg.det(a)
print(det)
```

inverse of matrix

```python
inv=np.linalg.inv(a)
print(inv)
```

Transpose

```python
trpose=np.linalg.matrix_transpose(p)
print(trpose)
```
