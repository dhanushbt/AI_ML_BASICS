```python
import numpy as np
np.set_printoptions(precision=4,suppress=True)
```

```python
a=np.array([[1,2,3,4],[1,1,2,3],[0,1,1,0]])
```

```python
a
```

Finding SVD should look like

**[3 * 3] [diagonal values][ 4 * 4]**
rank is 3 so atmost 3 diagonal values


```python
u,v,s=np.linalg.svd(a,full_matrices=True)
```

```python
v
```

```python
u.dot(u.T)
```

create diagonal matrix

```python
sd=np.diag(v)
```

```python
sd
```

```python
b=np.zeros((3,4))
```

```python
b[:,:-1]=sd
```

```python
b
```

```python
sigma=b
```

```python
print(np.dot(np.dot(u,sigma),s))
```
